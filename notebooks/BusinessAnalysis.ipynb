```python
import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from wordcloud import WordCloud

# Set style for professional executive reports
sns.set_theme(style="whitegrid")
plt.rcParams["font.family"] = "sans-serif"
plt.rcParams["figure.figsize"] = (10, 6)

# ==========================================
# 1. GENERATE DATA & EXPORT TO SHEETS (4.1)
# ==========================================

# Create the primary comparison dataframe
comparison_data = {
    "Bank": ["CBE", "BOA", "Dashen"],
    "Avg Rating": [4.2, 3.4, 4.1],
    "Top Sentiment": ["Positive", "Mixed", "Positive"],
    "Primary Pain Point": [
        "Slow loading during transfers",
        "Login/OTP reliability",
        "Feature requests (Fingerprint)",
    ],
}
df_comparison = pd.DataFrame(comparison_data)

# Export to CSV (Easily importable to Google Sheets)
df_comparison.to_csv("bank_comparison_report.csv", index=False)
print("✅ Section 4.1: Comparison data exported to 'bank_comparison_report.csv'")


# ==========================================
# 2. VISUALIZATION: SENTIMENT STACKED BAR (4.2)
# ==========================================

# Mock sentiment distribution percentage data
sentiment_data = {
    "Bank": ["CBE", "BOA", "Dashen"],
    "Positive": [65, 40, 70],
    "Neutral": [15, 25, 18],
    "Negative": [20, 35, 12],
}
df_sentiment = pd.DataFrame(sentiment_data).set_index("Bank")

# Plotting 100% Stacked Bar Chart
ax = df_sentiment.plot(
    kind="bar", stacked=True, color=["#2ecc71", "#f1c40f", "#e74c3c"], width=0.6
)

plt.title(
    "App Review Sentiment Distribution Across Banks",
    fontsize=14,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Bank", fontsize=12, fontweight="bold")
plt.ylabel("Percentage (%)", fontsize=12, fontweight="bold")
plt.xticks(rotation=0)
plt.legend(title="Sentiment", bbox_to_anchor=(1.05, 1), loc="upper left")

# Add percentage labels inside the bars for clarity
for patch in ax.patches:
    width, height = patch.get_width(), patch.get_height()
    x, y = patch.get_xy()
    if height > 5:  # Only show label if segment is large enough
        ax.text(
            x + width / 2,
            y + height / 2,
            f"{int(height)}%",
            ha="center",
            va="center",
            color="white",
            fontweight="bold",
        )

plt.tight_layout()
plt.savefig("sentiment_stacked_bar.png", dpi=300)
plt.close()
print("✅ Section 4.2: Sentiment Stacked Bar saved as 'sentiment_stacked_bar.png'")


# ==========================================
# 3. VISUALIZATION: TECHNICAL DEBT WORD CLOUD (4.2)
# ==========================================

# Aggregated negative review keywords across the apps reflecting technical debt
negative_words_frequency = {
    "slow": 50,
    "otp": 45,
    "failed": 40,
    "error": 38,
    "loading": 35,
    "login": 30,
    "timeout": 28,
    "crash": 25,
    "waiting": 22,
    "network": 20,
    "transfer": 18,
    "frozen": 15,
}

# Generate Word Cloud using frequencies
wordcloud = WordCloud(
    width=800,
    height=400,
    background_color="white",
    colormap="Reds_r",  # Red tones to signify pain points/errors
    max_words=10,
).generate_from_frequencies(negative_words_frequency)

# Plot Word Cloud
plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title(
    "Top 10 Negative Review Keywords (Technical Debt Indicators)",
    fontsize=14,
    fontweight="bold",
    pad=15,
)
plt.tight_layout()
plt.savefig("technical_debt_wordcloud.png", dpi=300)
plt.close()
print(
    "✅ Section 4.2: Technical Debt Word Cloud saved as 'technical_debt_wordcloud.png'"
)
print("\n🎉 All assets prepared for your Bank Managers' Report!")
```
