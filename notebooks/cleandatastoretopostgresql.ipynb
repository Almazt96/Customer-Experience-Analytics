```python
# Import Libraries
import psycopg2
import pandas as pd
from sqlalchemy import create_engine

# Database connection
# Connect to psycopg2 database (or create it if it doesn't exist)
from sqlalchemy import create_engine
connection = psycopg2.connect("dbname=mobileapp user=postgres password=Leul@123")
# Create a SQLAlchemy engine
engine = create_engine('postgresql+psycopg2://postgres:Leul@123@localhost/mobileapp')
print("✅ Database connection established.", engine)
# Query to fetch reviews
query = "SELECT review_text FROM reviews"   
print("✅ Connected to the database successfully.")

# Reviews Table
CREATE TABLE reviews (
    review_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    user_id VARCHAR(50) REFERENCES users(user_id) ON DELETE CASCADE,
    product_id VARCHAR(50) REFERENCES products(product_id) ON DELETE CASCADE,
    rating SMALLINT CHECK (rating >= 1 AND rating <= 5),
    review_title VARCHAR(255),
    review_text TEXT NOT NULL,
    review_date TIMESTAMP WITH TIME ZONE NOT NULL,
    helpful_count INTEGER DEFAULT 0,
    created_at TIMESTAMP WITH TIME ZONE DEFAULT CURRENT_TIMESTAMP
)
# reviews into a DataFrame    
df = pd.read_sql(query, engine)
print("✅ Fetched reviews into DataFrame.")
# Display the first few reviews
print(df.head())

```
